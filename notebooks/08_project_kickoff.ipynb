## 🏗️ Part 3: Architecture Design

Now let's design how your agent will work.

```python
# Architecture Planning
print("=" * 60)
print("ARCHITECTURE DESIGN")
print("=" * 60)
print()

architecture_guide = """
# ARCHITECTURE DESIGN GUIDE

## Component Breakdown

Break your agent into logical components:

1. INPUT HANDLER
   - What: Receives and validates user input
   - How: Parse user query, validate parameters
   - Example: "Get topic and max_papers from user"

2. CORE LOGIC
   - What: The main workflow of your agent
   - How: Orchestrate the steps, make decisions
   - Example: "Loop: search → filter → extract → compare"

3. TOOL LAYER (if using tools)
   - What: Interfaces to external services
   - How: MCP servers, API calls, file operations
   - Example: "arXiv search tool, PDF parser tool"

4. LLM INTERACTION LAYER
   - What: All LLM calls with proper prompting
   - How: CO-STAR prompts, structured outputs, CoT
   - Example: "Extract methodology using structured JSON output"

5. OUTPUT GENERATOR
   - What: Format and present results
   - How: Generate markdown, tables, visualizations
   - Example: "Create comparison table in markdown"

## Data Flow Diagram

Create a flow diagram for your agent:

┌─────────────┐
│ User Input  │
│ "Topic: X"  │
└──────┬──────┘
       │
       ▼
┌─────────────────┐
│  Search Papers  │  ← Tool/API Call
│  (arXiv API)    │
└──────┬──────────┘
       │
       ▼
┌─────────────────┐
│  Filter Top N   │  ← LLM: Relevance check
│  Most Relevant  │
└──────┬──────────┘
       │
       ▼
┌─────────────────┐
│ Extract Methods │  ← LLM: Structured extraction
│  (JSON output)  │
└──────┬──────────┘
       │
       ▼
┌─────────────────┐
│ Compare Methods │  ← LLM: CoT analysis
│  (find patterns)│
└──────┬──────────┘
       │
       ▼
┌─────────────────┐
│ Generate Report │  ← Format output
│  (Markdown)     │
└──────┬──────────┘
       │
       ▼
┌─────────────────┐
│ Return to User  │
└─────────────────┘

## Module Structure

Organize your code into modules:

project/
├── main.py (or main.ipynb)           # Entry point
├── config.py                          # Configuration
├── agents/
│   ├── __init__.py
│   ├── research_agent.py             # Main agent class
│   └── workflow.py                   # Workflow logic
├── tools/
│   ├── __init__.py
│   ├── paper_search.py               # Search functionality
│   ├── pdf_parser.py                 # PDF processing
│   └── formatter.py                  # Output formatting
├── prompts/
│   ├── __init__.py
│   ├── extraction.py                 # Extraction prompts
│   └── analysis.py                   # Analysis prompts
└── outputs/
    └── [generated files]

## Prompt Library

Document your key prompts:

PROMPT_EXTRACT_METHOD = '''
{CO-STAR structure}
Extract the methodology from this paper...
'''

PROMPT_COMPARE = '''
{CO-STAR structure}
Compare these methodologies...
'''

## Error Handling Strategy

Plan for failures:

1. Input Validation
   - Check: Valid topic? Reasonable parameters?
   - Action: Return clear error message

2. API Failures
   - Check: Rate limits? Network issues?
   - Action: Retry with backoff, use cache, fail gracefully

3. LLM Failures
   - Check: Invalid JSON? Poor quality output?
   - Action: Retry with better prompt, use fallback

4. Partial Success
   - Check: Got some but not all results?
   - Action: Return partial results with warning
"""

print(architecture_guide)
print()

# Save architecture guide
arch_path = os.path.join(parent_dir, 'outputs', 'architecture_guide.md')

with open(arch_path, 'w') as f:
    f.write(architecture_guide)

print(f"💾 Architecture guide saved to: {arch_path}")
print()
```

### 📝 Task 2: Design Your Architecture

**Goal:** Create a detailed architecture for your agent.

```python
# TODO - Task 2: Architecture Design
print("=" * 60)
print("TASK 2: Design Your Architecture")
print("=" * 60)
print()

# ============================================================================
# TODO: Design your agent architecture
# ============================================================================

my_architecture = """
# MY AGENT ARCHITECTURE

## Component Breakdown

### 1. INPUT HANDLER
Purpose: Capture and preprocess user natural language queries
Implementation:
- Simple CLI or API endpoint (FastAPI)
- Clean input (strip spaces, basic validation)
- Optionally detect intent (aggregation, filtering, grouping)

### 2. CORE WORKFLOW
Purpose: Convert natural language into database results

Steps:
  1. Receive user query
  2. Inject database schema into prompt
  3. Generate SQL query using LLM
  4. Validate SQL (SELECT-only, schema check)
  5. Execute SQL query on database
  6. Format results (table + optional summary)
  7. Return output to user

### 3. TOOL LAYER (if applicable)
Tools needed:
  - Database Connector (SQLAlchemy / sqlite3): Execute SQL queries
  - Schema Loader: Retrieve table/column metadata
  - SQL Validator: Ensure safe queries (block DELETE/UPDATE/DROP)
  - Pandas: Format query results into tables

### 4. LLM INTERACTION
LLM calls needed:

  Call 1: Natural Language → SQL generation  
    → Model: GPT-4.1 (or equivalent)  
    → Output format: JSON  
    Example:
    {
      "sql": "...",
      "explanation": "..."
    }

  Call 2: Result summarization (optional)  
    → Model: lighter model (GPT-4o-mini or similar)  
    → Output format: text summary  

### 5. OUTPUT GENERATOR
Output format: JSON + Markdown

Structure:
- SQL query used
- Table result (DataFrame → Markdown table)
- Optional summary
- Optional metadata (execution time, row count)

Example:
{
  "query": "...",
  "sql": "...",
  "result_table": "...",
  "summary": "..."
}

## Data Flow

User Input 
   ↓
Input Handler
   ↓
Schema Injection
   ↓
LLM (Generate SQL)
   ↓
SQL Validation
   ↓
Database Execution
   ↓
Result Formatting (pandas)
   ↓
LLM (Optional Summary)
   ↓
Final Output

## File Structure

nl_sql_agent/
  ├── main.py
  ├── config.py
  ├── db/
  │   ├── connection.py
  │   └── schema.py
  ├── agents/
  │   └── sql_agent.py
  ├── prompts/
  │   └── sql_generation_prompt.py
  ├── utils/
  │   ├── validator.py
  │   └── formatter.py
  └── outputs/

## Key Prompts

EXTRACTION_PROMPT:
'''
You are a data analyst expert.

Context:
You are given a database schema and a user question.

Schema:
{schema}

Task:
Convert the user's question into a valid SQL query.

Constraints:
- Only generate SELECT queries
- Use only tables and columns from the schema
- Avoid unnecessary complexity
- Include aggregation if needed (SUM, COUNT, etc.)

Output format (JSON):
{
  "sql": "your SQL query",
  "explanation": "brief explanation"
}

User Question:
{user_query}
'''

ANALYSIS_PROMPT:
'''
You are a business analyst.

Given the following query result:
{table}

Summarize key insights in 2-3 sentences.
Focus on trends, comparisons, or anomalies.
'''

## Error Handling

1. Input errors:
   - Empty or unclear query → return clarification request

2. API failures:
   - Retry LLM call (max 2 retries)
   - Fallback to simpler prompt

3. LLM failures:
   - Validate JSON output
   - If invalid → re-prompt with stricter instructions

4. Partial results:
   - If SQL runs but result is empty → return message + possible reasons
   - If schema mismatch → suggest corrected columns
"""

print(my_architecture)
print()

# Save architecture
arch_doc_path = os.path.join(parent_dir, 'outputs', 'my_architecture.md')

with open(arch_doc_path, 'w') as f:
    f.write(f"# My Agent Architecture\n")
    f.write(f"**Created:** {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n\n")
    f.write(my_architecture)

print(f"💾 Your architecture saved to: {arch_doc_path}")
print()

# ========================================================================
# TODO: Validate architecture with Claude
# ========================================================================

print("=" * 60)
print("ARCHITECTURE VALIDATION")
print("=" * 60)
print()

validation_prompt = f"""Review this agent architecture and identify potential issues:

{my_architecture}

Check for:
1. Missing components
2. Unclear data flow
3. Potential bottlenecks
4. Error handling gaps
5. Scalability concerns

Provide specific, actionable feedback."""

print("Validating architecture with Claude...")
print()

response = client.generate(
    prompt=validation_prompt,
    temperature=0.3,
    max_tokens=500
)

if "error" not in response:
    print("Architecture Review:")
    print("=" * 60)
    print(response['content'])
    print("=" * 60)
    tracker.add_call(response)
    
    # Save review
    with open(arch_doc_path, 'a') as f:
        f.write("\n\n---\n\n")
        f.write("## Architecture Review\n\n")
        f.write(response['content'])
    
    print()
    print("💾 Review added to your architecture document")
else:
    print(f"Error: {response['error']}")

print()

# ========================================================================
# TODO: Reflection
# ========================================================================

print("=" * 60)
print("REFLECTION")
print("=" * 60)
print()

reflection = """
### Architecture Clarity

**Can you explain your architecture to someone else?** Yes

The flow is straightforward: user query → LLM generates SQL → validate → execute → return results. Each component has a clear role and the data flow is linear.

### Complexity Assessment

**Is your architecture too complex?** No

The design is modular but still simple. Each component has a single responsibility.

**Is it too simple?** Slightly

It currently lacks:
- Advanced query optimization
- Handling ambiguous queries
- Multi-step reasoning for complex analytics

### Component Dependencies

**Which component is most critical?**
LLM SQL Generation

This is the core of the system. If SQL generation fails, the entire pipeline breaks, even if all other components work perfectly.

**Which component is riskiest?**
SQL Validation + Schema Alignment

Because:
- LLM may hallucinate columns/tables
- Incorrect validation can either block valid queries or allow unsafe ones

### LLM Usage

**Total LLM calls per workflow run:** 1–2

**Is this reasonable?** Yes

- 1 call for SQL generation
- Optional 1 call for summarization

Cost and latency are low enough for real-time usage.

**Could you reduce calls without losing quality?**
Yes

- Make summarization optional or rule-based (using pandas stats)
- Combine SQL + explanation in one call (already done)

### Tool Selection

**Which tools/APIs are you most uncertain about?**
- SQL validation libraries (accuracy vs strictness)
- Handling complex schemas with many joins

**Do you have backups if a tool fails?**
Yes

- If LLM fails → retry with simplified prompt
- If SQL fails → return error + suggested correction
- If validation fails → fallback to stricter rule-based filtering

### Code Organization

**How will you organize your code?**
- Separate modules:
  - input handling
  - LLM interaction
  - database access
  - validation
  - formatting

**Is it testable?**
Yes

- Unit tests:
  - SQL generation (mock LLM output)
  - SQL validation rules
  - DB execution (test database)
- Integration tests:
  - End-to-end query → result pipeline

### Architecture Feedback

**Most valuable insight from the review:**
The importance of enforcing strict schema-aware prompting and validation to prevent hallucinated SQL.

**What will you change?**
- Add stronger schema injection into prompts
- Implement stricter SQL validation rules
- Add logging for debugging LLM outputs

### Implementation Confidence

**How confident are you in implementing this?** 4/5

Confidence is high due to familiarity with Python, databases, and APIs. Some uncertainty remains around edge cases and complex queries.

### Biggest Architecture Challenge

Handling ambiguous or complex natural language queries (e.g., multi-table joins, unclear metrics)

How will you tackle it?
- Start with simple queries (MVP)
- Gradually add support for joins and aggregations
- Improve prompts with examples (few-shot learning)
- Add user clarification step when needed
"""

print(reflection)

append_to_reflection(
    notebook="08",
    section_title="Task 2 - Architecture Design",
    reflection_content=reflection,
    output_dir=os.path.join(parent_dir, 'outputs')
)

print()
print("💾 Reflection saved to outputs/homework_reflection.md")
print()
```
